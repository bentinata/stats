# Nonlinear Relationship & Transformation (Simple Version)

| $x$ | -8 | -16 | 12 | 2 | 18 |
| :--- | :---: | :---: | :---: | :---: | :---: |
| $y$ | 58 | 247 | 153 | 3 | 341 |

## a. Create a scatter diagram and describe the relationship between x and y.

```python
import matplotlib.pyplot as plt

x = [-8, -16, 12, 2, 18]
y = [58, 247, 153, 3, 341]

plt.figure(figsize=(6, 4))
plt.scatter(x, y, color="#1f77b4")
plt.title("x vs. y")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(-20, 20)
plt.ylim(0, 400)
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

**Relationship:** Nonlinear / curvilinear (U-shaped parabolic relationship where $y \approx x^2$).

## b. Compute the correlation coefficient.

$$r = \frac{2280.8}{\sqrt{779.2 \times 75431.2}} \approx \mathbf{0.298}$$

Weak linear correlation ($r \approx 0.30$) due to the U-shaped nonlinear curve.

```python
import numpy as np

r_xy = np.corrcoef(x, y)[0, 1]
print(f"Correlation coefficient r(x, y) = {r_xy:.4f}")
```

## c. Transform the x variable by squaring each value, x2.

Values of $x^2$: **64, 256, 144, 4, 324**

```python
x2 = [val**2 for val in x]
print(f"x^2 = {x2}")
```

## d. Create a scatter diagram and describe the relationship between x2 and y.

```python
plt.figure(figsize=(6, 4))
plt.scatter(x2, y, color="#1f77b4")
plt.title("x² vs. y")
plt.xlabel("x²")
plt.ylabel("y")
plt.xlim(0, 400)
plt.ylim(0, 400)
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

**Relationship:** Very strong, positive linear relationship.

## e. Compute the correlation coefficient between x2 and y.

$$r = \frac{72435.2}{\sqrt{69907.2 \times 75431.2}} \approx \mathbf{0.998}$$

Nearly perfect positive linear relationship ($r \approx 0.998$, $r^2 \approx 0.995$).

```python
r_x2y = np.corrcoef(x2, y)[0, 1]
print(f"Correlation coefficient r(x^2, y) = {r_x2y:.4f}")
```

## f. Compare the relationships between x and y, and x2 and y.

- $x$ vs. $y$: Nonlinear (U-shaped) curve, low linear correlation ($r = 0.298$).
- $x^2$ vs. $y$: Linear straight line, extremely high correlation ($r = 0.998$).
- Squaring $x$ linearizes the quadratic relationship.

## g. Interpret your results.

1. Pearson's correlation coefficient only evaluates **linear** relationships.
2. A low linear correlation does not mean no relationship exists; always check a scatter plot.
3. Transforming variables ($x \to x^2$) can convert a nonlinear relationship into a strong linear one.

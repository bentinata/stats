# Digital Camera Sales & TV Commercial Airings (Simple Version)

| Location    | Airings ($x$) | Sales ($ thousands) ($y$) |
| :---------- | :------------ | :------------------------ |
| Providence  | 4             | 15                        |
| Springfield | 2             | 8                         |
| New Haven   | 5             | 21                        |
| Boston      | 6             | 24                        |
| Hartford    | 3             | 17                        |

## a. What is the dependent variable?

- **Dependent variable ($y$):** Digital camera sales ($ thousands)
- **Independent variable ($x$):** Number of commercial airings

## b. Draw a scatter diagram.

```python
import matplotlib.pyplot as plt

airings = [4, 2, 5, 6, 3]
sales = [15, 8, 21, 24, 17]

plt.figure(figsize=(6, 4))
plt.scatter(airings, sales, color="#1f77b4")
plt.title("Airings vs. Sales")
plt.xlabel("Number of Airings")
plt.ylabel("Sales ($ thousands)")
plt.xlim(1, 7)
plt.ylim(5, 25)
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

## c. Determine the correlation coefficient.

$$r = \frac{SS_{xy}}{\sqrt{SS_{xx} \cdot SS_{yy}}} = \frac{36}{\sqrt{10 \times 150}} = \frac{36}{\sqrt{1500}} \approx 0.930$$

```python
import numpy as np

r = np.corrcoef(airings, sales)[0, 1]
print(f"Correlation coefficient r = {r:.4f}")
```

## d. Interpret these statistical measures.

There is a very strong positive linear relationship ($r \approx 0.93$) between TV commercial airings and digital camera sales.
